```python
%sql
--Crear el Catálogo (Contenedor principal)
CREATE CATALOG IF NOT EXISTS proyectos_bigdata; 

--Crear el Esquema (Base de Datos)
CREATE SCHEMA IF NOT EXISTS proyectos_bigdata.powerlifting_romen; 

--Usar el Esquema
USE proyectos_bigdata.powerlifting_romen;
```

```python
%sql
--DDL para la tabla de Encuentros (meets.csv)
CREATE TABLE IF NOT EXISTS proyectos_bigdata.powerlifting_romen.tbl_encuentro (
    MeetID LONG NOT NULL COMMENT "Llave primaria",
    MeetPath STRING,
    Federation STRING,
    Date STRING,
    MeetCountry STRING,
    MeetState STRING,
    MeetTown  STRING,
    MeetName STRING,
    PRIMARY KEY (MeetID)
);
```

```python
%sql
SHOW CREATE TABLE proyectos_bigdata.powerlifting_romen.tbl_encuentro;
```

```python
%sql
DESCRIBE TABLE proyectos_bigdata.powerlifting_romen.tbl_encuentro;
```

```python
%sql
SELECT * FROM proyectos_bigdata.powerlifting_romen.tbl_encuentro
LIMIT 5
```

```python
%sql
--DDL para la tabla de Competidores (openpowerlifting.csv)
CREATE TABLE IF NOT EXISTS proyectos_bigdata.powerlifting_romen.tbl_competidor_powerlifting (
    MeetIDCompetidor LONG NOT NULL COMMENT "Llave primaria",
    Name STRING NOT NULL,
    Sex STRING,
    Equipment STRING,
    Age INT,
    Division STRING,
    BodyweightKg LONG,
    WeightClassKg INT,
    Squat4Kg INT,
    BestSquatKg INT, 
    Bench4Kg INT,
    BestBenchKg INT,
    Deadlift4Kg INT,
    BestDeadliftKg INT,
    TotalKg LONG,
    Place INT, 
    Wilks LONG,
    PRIMARY KEY (MeetIDCompetidor)
);
```

```python
%sql
DESCRIBE TABLE proyectos_bigdata.powerlifting_romen.tbl_competidor_powerlifting;
```

```python
%sql
SELECT * FROM proyectos_bigdata.powerlifting_romen.tbl_competidor_powerlifting
LIMIT 5
```

```python
%sql
-- DDL para la tabla de Hechos (Central de Resultados)
CREATE TABLE IF NOT EXISTS proyectos_bigdata.powerlifting_romen.tbl_encuentro_competidor_powerliftings (
    MeetID LONG NOT NULL,
    MeetIDCompetidor LONG NOT NULL,
    Name STRING NOT NULL,
    Sex STRING,
    Equipment STRING,
    Age INT,
    Division STRING,
    BodyweightKg LONG,
    WeightClassKg INT,
    Squat4Kg INT,
    BestSquatKg INT, 
    Bench4Kg INT,
    BestBenchKg INT,
    Deadlift4Kg INT,
    BestDeadliftKg INT,
    TotalKg LONG,
    Place INT, 
    Wilks LONG,
    CONSTRAINT pk_encuentro_competidor_powerliftings PRIMARY KEY (MeetID, MeetIDCompetidor)
)
```

```python
%sql
INSERT INTO proyectos_bigdata.powerlifting_romen.tbl_encuentro_competidor_powerliftings (
    MeetIDCompetidor, MeetID, Name, Sex, Age, Place, Wilks
)
SELECT
    1 AS MeetIDCompetidor,
    1 AS MeetID,
    Name,
    Sex,
    Age,
    try_cast(Place AS INT) AS Place,
    Wilks
FROM 
    proyectos_bigdata.powerlifting_romen.tbl_competidor_powerlifting;
```
